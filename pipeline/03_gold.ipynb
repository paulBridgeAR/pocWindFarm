```python
from pyspark import pipelines as dp
from nb_transforms import daily_stats, find_anomalies

 
@dp.materialized_view(
    name="gold_turbine_daily_stats",
    comment="Min, max and average power per turbine per day, with completeness.",
    table_properties={"quality": "gold"},
)
def gold_turbine_daily_stats():
    return daily_stats(spark.read.table("silver_turbine_readings"))
 
 
@dp.materialized_view(
    name="gold_turbine_anomalies",
    comment="Turbine-days more than 2 sd from the fleet mean for that day.",
    table_properties={"quality": "gold"},
)
def gold_turbine_anomalies():
    return find_anomalies(spark.read.table("gold_turbine_daily_stats"))
 
```
